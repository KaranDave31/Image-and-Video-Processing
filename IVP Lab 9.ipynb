```python
import cv2
import numpy as np
```

```python
image = cv2.imread('pic.jpg')
cv2.imshow('Original', image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
gray_image = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
img = cv2.resize(gray_image,(512,512))
cv2.imshow('Resized',img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
seed_x = int(input("Enter the x-coordinate of the seed pixel: "))
seed_y = int(input("Enter the y-coordinate of the seed pixel: "))
seed_point = (seed_x, seed_y)
```

```python
_, binary_image = cv2.threshold(gray_image, 127, 255, cv2.THRESH_BINARY)
```

```python
def flood_fill(image, seed_point, new_color):
    h, w = image.shape
    fill_image = image.copy()
    target_color = fill_image[seed_point[1], seed_point[0]]

    stack = [seed_point]
    
    while stack:
        x, y = stack.pop()
        if (0 <= x < w) and (0 <= y < h) and fill_image[y, x] == target_color and fill_image[y, x] != new_color:
            fill_image[y, x] = new_color
            
            stack.append((x + 1, y))  
            stack.append((x - 1, y))  
            stack.append((x, y + 1))  
            stack.append((x, y - 1))  
    
    return fill_image
```

```python
new_color = 255  # For white
filled_image = flood_fill(binary_image, seed_point, new_color)
```

```python
cv2.imshow('Original Image', gray_image)
cv2.imshow('Flood Fill Result', filled_image)

cv2.waitKey(0)  # Wait for a key press
cv2.destroyAllWindows()
```

```python
def region_growing(image, seed, threshold=20):
    output = np.zeros_like(image)
    h, w = image.shape
    visited = np.zeros((h, w), dtype=bool)
    pixels_to_process = [seed]
    seed_value = image[seed]

    while pixels_to_process:
        x, y = pixels_to_process.pop(0)
        
        # Check boundaries and whether the pixel has been visited
        if x < 0 or x >= h or y < 0 or y >= w or visited[x, y]:
            continue
        
        # Check if the pixel value is within the threshold
        if abs(int(image[x, y]) - int(seed_value)) <= threshold:
            visited[x, y] = True
            output[x, y] = 255  # Set output pixel to white
            
            # Add neighboring pixels (4-connectivity)
            pixels_to_process.append((x-1, y))  # Up
            pixels_to_process.append((x+1, y))  # Down
            pixels_to_process.append((x, y-1))  # Left
            pixels_to_process.append((x, y+1))  # Right
            
    return output
```

```python
seed_pixel = (150, 150)
```

```python
output_image = region_growing(binary_image, seed_pixel)

```

```python
# Step 5: Display the original and output images using OpenCV
cv2.imshow('Original Image', gray_image)
cv2.imshow('Region Growing Output', output_image)

cv2.waitKey(0)  # Wait for a key press
cv2.destroyAllWindows()
```
